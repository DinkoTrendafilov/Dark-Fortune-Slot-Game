# 🎰 DARK FORTUNE: SLOT MACHINE (40 LINES)

## Software Design Document (SDD)

---

## 1. Executive Summary

### Project Overview

Dark Fortune is a **slot machine** designed with 5 reels, 4 rows, and **40 optimized paylines**. The game features 30 dark-themed symbols and is mathematically validated through exhaustive combinatorial analysis and Monte Carlo simulations.

### Key Achievement

The project successfully implements and validates a slot machine with **96.0186% theoretical RTP** and **1075.90% Volatility Index (CV)**. The simulation of **10,000,000 spins** confirms the mathematical model with **only 0.0227% RTP deviation**. With **99.16% hit rate**, the game delivers frequent wins and engaging gameplay.

---

## 2. System Architecture

### Core Components

The slot machine consists of five primary modules:

1. **Combinatorial Engine** – analyzes all 24,300,000 possible reel combinations (30 symbols × 5 reels)
2. **Game Logic** – handles spin generation, payline evaluation, and win calculation
3. **Monte Carlo Simulator** – executes millions of random spins for statistical validation
4. **Statistical Analyzer** – computes RTP, hit rate, volatility, VaR, and survival metrics
5. **Bankroll Simulator** – models player survival across 100–25,000 spin sessions

### Technical Stack

| Component | Technology |
|-----------|------------|
| Core Language | Python 3.9+ |
| Numerical Computing | NumPy |
| Statistical Analysis | SciPy (Beta) |
| Visualization | Matplotlib |
| Data Structures | Collections (Counter) |

## 4. Paytable Design

### The Paytable

| Match Type | Coefficient | RTP Contribution |
|------------|-------------|------------------|
| 1 match | 0.4× (0.3 Fund + 0.1 Jackpot) | 28.1493% |
| 2 matches | 1.3× | 37.1415% |
| 3 matches | 21× | 21.8037% |
| 4 matches | Pick-Em (270–2,236×) | 7.6895% |
| 5 matches | Pick-Em (6,290–52,000×) | 1.2345% |
| **Full Screen** | **10,000,000× per line** | **Included in RTP via simulation** |

### Design Philosophy

The design prioritizes:

- **Fund System** — 1-match contributes 0.3× to fund
- **Jackpot System** — 1-match contributes 0.1× to jackpot with 1/10,000 hit chance
- **Wheel Bonus** — triggered at 100 empty lines (5 attempts, ~67.2% win rate)
- **Pick-Em Bonus** — hypergeometric distribution (N=16, K=8, n=4)
- **Full Screen Bonus** — 10,000,000× per line if all 20 cells match
- **High Hit Rate** — 99.16% of spins produce a win

### RTP Calculation

**Theoretical RTP:** 96.0186%
**Simulated RTP:** 95.9959% (from 10,000,000 spins)
**Deviation:** Only 0.0227% — excellent validation

## 5. Combinatorial Analysis

### Distribution of Outcomes

| Max Pairs | Count | Probability | 1 in |

| 1 | 17,100,720 | 70.3733% | 1.4 |
| 2 | 6,942,600 | 28.5704% | 3.5 |
| 3 | 252,300 | 1.0383% | 96.3 |
| 4 | 4,350 | 0.0179% | 5,586 |
| 5 | 30 | 0.0001% | 810,000 |

### Key Insights

- **70.37% of lines have no win** — this is normal for slot machines and feeds the fund system
- **2-matches occur in 28.57% of combinations** — the primary source of frequent wins
- **5-matches (jackpot) occur only 30 times** out of 24.3M combinations
- **Full Screen Bonus** is extremely rare — 0 occurrences in 10M spins confirms the 10,000,000× multiplier

## 13. Bankroll Survival Analysis

### Methodology

We simulated **1,000 sessions** for each spin target, starting with **100 credits** and betting **1 credit per spin** (0.025 per line × 40 lines). A session ends early if the balance falls below the bet amount.

### Survival Rates

| Spins per Session | Survival Rate | Bust Rate | Avg Balance |
|-------------------|---------------|-----------|-------------|
| 100 | 100.00% | 0.00% | 90 |
| 300 | 100.00% | 0.00% | 69 |
| 500 | 83.30% | 16.70% | 45 |
| 1,000 | 23.70% | 76.30% | 21 |
| 2,500 | 3.70% | 96.30% | 8 |
| 5,000 | 0.80% | 99.20% | 3 |
| 25,000 | 0.10% | 99.90% | 1 |

### Key Insights

- **100% survival for the first 300 spins**, showcasing the high hit rate
- **Survival drops sharply after 400 spins**, indicating the effect of volatility
- The game is **highly volatile** — players either bust or survive with large balances

## 14. Risk Metrics

### Volatility Index (CV)

| Type | Value |
|------|-------|
| **Coefficient of Variation** | 1075.90% |

This exceptionally high CV is a mathematical artifact of the game's high hit rate and small average win. The standard deviation (10.33) is over ten times larger than the average win (0.96), making the game highly volatile in terms of win size variation despite the frequent wins.

### Value at Risk (VaR)

| Confidence Level | VaR (credits) |
|------------------|---------------|
| 95% | 0.97 credits |

### Interpretation

At 95% confidence, the maximum loss on a single spin is 0.97 credits. This is typical for low-volatility games with 40 lines.

## 16. Player Experience Design

### Target Player Profile

- **Casual players** seeking entertainment over profit
- **Extended sessions** (500+ spins) with many small wins
- **Low risk tolerance** — wants frequent wins, not big swings
- **Bonus chasers** — dreams of 10,000,000× Full Screen or 52,000× Pick-Em wins

### Gameplay Experience

| Phase | Experience |
|-------|------------|
| First 300 spins | 100% survival, many small wins |
| 300–500 spins | Survival drops to 83.3%, still frequent wins |
| 500–1,000 spins | Survival drops to 23.7%, bust rate increases |
| 1,000+ spins | Majority of players bust, few survive with large balances |

### Design Strengths

- **Frequent wins** — 99.16% hit rate
- **Pick-Em bonus** — strategic element with huge payouts
- **Wheel bonus** — fund system adds depth
- **Full Screen potential** — 10,000,000× wins possible
- **Jackpot system** — adds excitement with random hits
- **Mathematically validated** — all metrics align with theory
